# Mutual Fund Analysis

Clean → score → rank top 30 low-risk / high-return schemes.

This notebook mirrors `scripts/run_pipeline.py` for interactive exploration.

```python
import sys
from pathlib import Path

import pandas as pd

ROOT = Path("..").resolve()
sys.path.insert(0, str(ROOT / "src"))

from mf_analysis.clean import clean_funds
from mf_analysis.insights import build_insights, summarize_universe
from mf_analysis.score import rank_funds, score_funds

raw = pd.read_csv(ROOT / "data" / "raw" / "mutual_funds_sample.csv")
raw.head()
```

```python
cleaned = clean_funds(raw)
cleaned.describe(include="all").T.head(12)
```

```python
scored = score_funds(cleaned)
top30 = rank_funds(scored, top_n=30)
top30[
    [
        "rank",
        "scheme_name",
        "fund_type",
        "return_3y_pct",
        "expense_ratio_pct",
        "composite_score",
    ]
]
```

```python
summary = summarize_universe(cleaned)
insights = build_insights(cleaned, top30)
pd.DataFrame(insights)
```
